##Tema: Classificação de riscos e estimativa de eventos climáticos extremos
Na Classificação (Nível de Risco):
O que preve: O Nível de Risco de Tempestade/Inundação no dia seguinte.

Na Regressão (Estimativa Contínua):O que preve: O volume exato de chuva do dia seguinte (Rainfall em $mm$).

```python
import pandas as pd
from google.colab import files

uploaded = files.upload()

df = pd.read_csv('weatherAUS.csv')
```

```python
print("=== Quantidade de valores nulos ===")
print(df.isnull().sum())
print("=== Primeiras Linhas ===")
print(df.head())
print(df.shape)
```

```python
# ==============================================================================
# TRATAMENTO DE NULOS E LIMPEZA INICIAL
# ==============================================================================

# 1. Remover colunas com excesso de nulos e identificadores (nomes exatos do seu print)
cols_para_remover = ['Evaporation', 'Sunshine', 'Cloud9am', 'Cloud3pm', 'Date', 'Location']
df = df.drop(columns=cols_para_remover, errors='ignore')

# 2. Remover linhas onde o alvo está vazio
df = df.dropna(subset=['RainTomorrow', 'Rainfall']).reset_index(drop=True)

# 3. Preenchimento de nulos nas colunas restantes
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

# Imputação pela Mediana (Numéricas) e Moda (Categóricas)
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("✅ Limpeza concluída!")
print(f"📊 Formato atual do dataset: {df.shape}")
print("Colunas mantidas:", df.columns.tolist())
```
